# Set up API Key

```python
from openai import OpenAI
import os
from dotenv import load_dotenv

# Load environment variables from a .env file
load_dotenv()

client = OpenAI(
    api_key=os.environ.get("OPENAI_API_KEY"),
)
```

# Prompt Engineering

## Zero-Shot Prompting

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": "What is the best marketing technique for tech companies?"},  
    ]
)
```

```python
print(response.choices[0].message.content)
```

# Streaming Responses

```python
stream = client.chat.completions.create(
    model="gpt-4o-mini", 
    messages=[
        {'role': 'user', 'content': "What is AI?"}
    ], 
    stream=True
)

for chunk in stream:
    print(chunk.choices[0].delta.content, end='', flush=True)
```

# Prompting Template

```python
def advanced_prompt(my_prompt, input=""):
    response = client.chat.completions.create(
        model="gpt-4o",
        messages=[
            {
                "role": "user",
                "content": [
                    {
                        "type": "text",
                        "text": my_prompt.format(text=input) if len(input) > 0 else my_prompt
                    },
                ],
            }
        ],
        max_tokens=500,
        temperature=0.7
    )
    print(response.choices[0].message.content)
    return response
```

```python
my_prompt = """
Create a list of three fictional book titles along with their authors and genres.
Provide it in JSON format with the following keys: book_id, title, author, genre.
"""

advanced_prompt(my_prompt)
```

```python
input = """
To make delicious and tasty brownies, start by preparing the ingredients: 200 grams of chocolate, 150 grams of all-purpose flour, 3 eggs, and 100 grams of butter. 
Preheat your oven to 175°C (350°F). Melt the chocolate and butter together until smooth. 
In a separate bowl, beat the eggs and gradually add the flour. 
Combine all ingredients and stir until the batter is well mixed. 
Pour the batter into a greased or lined baking pan and bake for 25 to 30 minutes, or until a toothpick inserted into the center comes out clean. 
Allow the brownies to cool before cutting them into squares. Serve and enjoy.
"""

my_prompt = """
You will be given a text enclosed in triple quotes.
If the text contains a series of instructions, rewrite those instructions in the following format:

Step 1 - …
Step 2 - …
…
Step N - …

If the text does not contain a series of instructions, simply write "No steps provided."

\"\"\"{text}\"\"\"
"""

advanced_prompt(my_prompt, input)
```
