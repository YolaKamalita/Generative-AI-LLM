# Set up API Key

```python
from openai import OpenAI
import os
from dotenv import load_dotenv

# Load environment variables from a .env file
load_dotenv()

client = OpenAI(
    api_key=os.environ.get("OPENAI_API_KEY"),
)
```

# Prompt Engineering

## Zero-Shot Prompting

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": "What is the best marketing technique for tech companies?"},  
    ]
)
```

```python
print(response.choices[0].message.content)
```

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": "What is the best marketing technique for tech companies?"},  
    ],
    temperature=1,
    top_p=1,
    # max_tokens=300,
    frequency_penalty=0,
)
```

```python
print(response.choices[0].message.content)
```

## One-Shot Prompting

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": "Example: What is This is April?"},
        {"role": "assistant", "content": "This is April is a retail company from Indonesia."},
        {"role": "user", "content": "What is H&M?"}
    ],
    temperature=0.1
)
```

```python
print(response.choices[0].message.content)
```

## Few-Shot Prompting

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": "Example: What is This is April?"},
        {"role": "assistant", "content": "This is April is a retail company from Indonesia."},
        {"role": "user", "content": "Example: What is H&M?"},
        {"role": "assistant", "content": "H&M is a retail company from Swedia."},
        {"role": "user", "content": "Example: What is Mango?"},
        {"role": "assistant", "content": "Mango is a retail company from Spain."},
        {"role": "user", "content": "What is Adidas?"},
    ],
    temperature=0.1
)
```

```python
print(response.choices[0].message.content)
```

Notes:

How to make it answer like the given examples?

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "Answer concisely like in the examples. Stick to the pattern: '[Brand] is a retail company from [Country].'"},
        {"role": "user", "content": "Example: What is This is April?"},
        {"role": "assistant", "content": "This is April is a retail company from Indonesia."},
        {"role": "user", "content": "Example: What is H&M?"},
        {"role": "assistant", "content": "H&M is a retail company from Swedia."},
        {"role": "user", "content": "Example: What is Mango?"},
        {"role": "assistant", "content": "Mango is a retail company from Spain."},
        {"role": "user", "content": "What is Adidas?"},
    ],
    temperature=0.1
)
```

```python
print(response.choices[0].message.content)
```
